## 3.List

```python
justice_league = [
    "Superman",
    "Batman",
    "Wonder Woman",
    "Flash",
    "Aquaman",
    "Green Lantern"
]

# 1. Number of members
print("Number of members:", len(justice_league))

# 2. Add Batgirl and Nightwing
justice_league.append("Batgirl")
justice_league.append("Nightwing")
print(justice_league)

# 3. Move Wonder Woman to beginning
justice_league.remove("Wonder Woman")
justice_league.insert(0, "Wonder Woman")
print(justice_league)

# 4. Move Green Lantern between Aquaman and Flash
justice_league.remove("Green Lantern")
aquaman_index = justice_league.index("Aquaman")
justice_league.insert(aquaman_index + 1, "Green Lantern")
print(justice_league)

# 5. Replace with new team
justice_league = [
    "Cyborg",
    "Shazam",
    "Hawkgirl",
    "Martian Manhunter",
    "Green Arrow"
]
print(justice_league)

# 6. Sort alphabetically
justice_league.sort()
print(justice_league)

# New leader
print("New Leader:", justice_league[0])
```

## 4.If Conditon

```python
height = float(input("Enter height in meters: "))
weight = float(input("Enter weight in kilograms: "))

bmi = weight / (height ** 2)

if bmi >= 30:
    print("Obesity")
elif bmi >= 25:
    print("Overweight")
elif bmi >= 18.5:
    print("Normal")
else:
    print("Underweight")
```

```python
Australia = ["Sydney", "Melbourne", "Brisbane", "Perth"]
UAE = ["Dubai", "Abu Dhabi", "Sharjah", "Ajman"]
India = ["Mumbai", "Bangalore", "Chennai", "Delhi"]

city = input("Enter a city name: ")

if city in Australia:
    print(city, "is in Australia")
elif city in UAE:
    print(city, "is in UAE")
elif city in India:
    print(city, "is in India")
else:
    print("City not found")
```

```python
Australia = ["Sydney", "Melbourne", "Brisbane", "Perth"]
UAE = ["Dubai", "Abu Dhabi", "Sharjah", "Ajman"]
India = ["Mumbai", "Bangalore", "Chennai", "Delhi"]

city1 = input("Enter the first city: ")
city2 = input("Enter the second city: ")

if city1 in Australia and city2 in Australia:
    print("Both cities are in Australia")
elif city1 in UAE and city2 in UAE:
    print("Both cities are in UAE")
elif city1 in India and city2 in India:
    print("Both cities are in India")
else:
    print("They don't belong to the same country")
```

## 5.For Loop

```python
import random

six_count = 0
one_count = 0
two_six_count = 0
previous = 0

for i in range(20):
    dice = random.randint(1, 6)
    print("Roll", i + 1, ":", dice)

    if dice == 6:
        six_count += 1

    if dice == 1:
        one_count += 1

    if dice == 6 and previous == 6:
        two_six_count += 1

    previous = dice

print("Number of 6s:", six_count)
print("Number of 1s:", one_count)
print("Two 6s in a row:", two_six_count)
```

```python
completed = 0

for i in range(10):
    completed += 10

    print("You completed", completed, "jumping jacks.")

    if completed == 100:
        print("Congratulations! You completed the workout.")
        break

    tired = input("Are you tired? ")

    if tired.lower() in ["yes", "y"]:
        skip = input("Do you want to skip the remaining sets? ")

        if skip.lower() in ["yes", "y"]:
            print("You completed a total of", completed, "jumping jacks.")
            break

    remaining = 100 - completed
    print(remaining, "jumping jacks remaining.")
```
